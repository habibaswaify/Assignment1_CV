```python
import numpy as np
import matplotlib.pyplot as plt
from starter_pack.code import utils
from filters_helpers import to_gray
from Hybrid_img import low_pass
```

```python
rings = utils.load_image("starter_pack\images\part2_frequencies\\rings.png")
rings_gray = to_gray(rings)
```

```python
utils.show(rings_gray)
```

```python
naive_down = rings_gray[::4, ::4]
sigmas = [1.0, 1.5, 2.0,2.25, 2.5, 3.0]
results = []

for s in sigmas:
    blurred = low_pass(rings_gray, sigma=s)
    results.append(blurred[::4, ::4])
```

```python
sweep_titles = [f"$\sigma = {s}$" for s in sigmas]
utils.show(*results, titles=sweep_titles)
```

```python
best_sigma = 2.5
best_idx = sigmas.index(best_sigma)
x = results[best_idx]
utils.show(
    naive_down, 
    x, 
    titles=[
        "Naive", 
        f"Blur ($\sigma={best_sigma}$)"
    ]
)
```
